```python
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Divar.csv", low_memory=False)
```

```python
df["cat2_slug"].value_counts()
```

```python
df["cat3_slug"].value_counts()
```

```python
# Plot data distribution for cat2
plt.figure(figsize=(8, 5))
df["cat2_slug"].value_counts().plot(kind="bar")
plt.title("Distribution of Ads by Level 2 Category")
plt.xlabel("Level 2 Category")
plt.ylabel("Number of Ads")
plt.xticks(rotation=45)
plt.show()
```

```python
# Plot data distribution for cat3
plt.figure(figsize=(10, 5))
df["cat3_slug"].value_counts().plot(kind="bar")
plt.title("Distribution of Ads by Level 3 Category")
plt.xlabel("Level 3 Category")
plt.ylabel("Number of Ads")
plt.xticks(rotation=90)
plt.show()
```
